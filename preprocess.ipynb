```python
import pandas as pd
import geopandas as gpd

u = pd.read_excel('data2/multiob-1.xlsx').dropna()
u.to_csv('CL/Ws.csv')

x = pd.read_excel('data2/data-1.xlsx').dropna()
```

```python
x.grid_code[~x.grid_code.isin(u.PAC)].unique()
```

```python
x = x[x.grid_code.isin(u.PAC)]
```

```python
x = gpd.GeoDataFrame(x, geometry=gpd.points_from_xy(x['centroid_x'], x['centroid_y'], crs=28413), crs=28413)
x['lu'] = 1
x['Q80'] = x['priority2'] > x['priority2'].quantile(.2)
x['Q65'] = x['priority2'] > x['priority2'].quantile(.35)
x['Q50'] = x['priority2'] > x['priority2'].quantile(.5)

x
```

```python

x.to_feather('CL/luX.feather')

x[x['Q80']].to_feather('CL/lu80.feather')
x[x['Q65']].to_feather('CL/lu65.feather')
x[x['Q50']].to_feather('CL/lu50.feather')
```
